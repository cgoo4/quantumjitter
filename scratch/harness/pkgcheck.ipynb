```r
dirs <- list.dirs("/Users/carlgoodwin/R Projects/quantumjitter/_site/site_libs/quarto-contrib/shinylive-0.10.12/shinylive/webr/packages", recursive = FALSE) |> basename()
print(setdiff(pkgs, dirs))   # metadata entries with no dir
print(setdiff(dirs, pkgs))   # dirs not in metadata
suspects <- c("ggfoundry", "usedthese", "grImport2", "jpeg", "png", "svglite", "kableExtra")
keep <- setdiff(pkgs, suspects)
db <- installed.packages()[, c("Package", "Depends", "Imports", "LinkingTo")]
deps_of_keep <- tools::package_dependencies(keep, db = db, which = c("Depends", "Imports", "LinkingTo"), recursive = TRUE)
closure <- unique(unlist(deps_of_keep))
print(setdiff(suspects, closure))          # suspects not needed by keep set
print(intersect(suspects, closure))        # suspects needed by keep set
```
